- Long Answer Questions

- Interview Questions
- MCQs (if applicable)

This enables students to efficiently prepare for examinations using only the relevant content retrieved from the knowledge base.

````python
from langchain_community.document_loaders import PyMuPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.tools import create_retriever_tool
from langgraph.prebuilt import create_react_agent
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage, SystemMessage


def important_questions_agent(query, pdf_path="/content/output.pdf"):

    # Load PDF
    loader = PyMuPDFLoader(pdf_path)
    docs = loader.load()

    # Split Documents
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=150
    )

    documents = splitter.split_documents(docs)

    # Embeddings
    embedding = HuggingFaceEmbeddings(
        model_name="BAAI/bge-small-en-v1.5"
    )

    # Vector Store
    vectorstore = FAISS.from_documents(
        documents,
        embedding
    )

    retriever = vectorstore.as_retriever(
        search_type="similarity",
        search_kwargs={"k":6}
    )

    # Tool
    rag_tool = create_retriever_tool(
        retriever=retriever,
        name="pdf_notes",
        description="Searches the uploaded study material and lecture notes."
    )

    # LLM
    llm = ChatGroq(
        model="llama-3.3-70b-versatile",
        temperature=0,
        api_key="YOUR KEY"
    )

    # ReAct Agent
    agent = create_react_agent(
        model=llm,
        tools=[rag_tool]
    )

    system_prompt = """
You are an expert Exam Preparation Assistant.

Your only knowledge source is the uploaded PDF.

Always use the pdf_notes tool.

If information is unavailable,
reply:
'I couldn't find this in the uploaded PDF.'

If the user asks for important questions:

Generate questions ONLY from the PDF.

Rank them by importance.

Return in this format:

# Important Questions

1.
2.
3.

If asked for MCQs:
Generate MCQs only from PDF.

If asked for Viva:
Generate viva questions only from PDF.

If asked for 2 marks:
Generate short questions.

If asked for 5 marks:
Generate descriptive questions.

Never use outside knowledge.
"""

    response = agent.invoke(
        {
            "messages": [
                SystemMessage(content=system_prompt),
                HumanMessage(content=query)
            ]
        }
    )

    return response["messages"][-1].content
````

````python
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_community.document_loaders import PyMuPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.tools import create_retriever_tool
from langgraph.prebuilt import create_react_agent
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_community.document_loaders import CSVLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.tools import create_retriever_tool
def placement_agent(query):

    loader = CSVLoader(
        file_path="/content/placement_data.csv",
        encoding="latin-1"
    )
    docs = loader.load()

    splitter = RecursiveCharacterTextSplitter(
        chunk_size=800,
        chunk_overlap=20,
        length_function=len,
        add_start_index=True,
    )
    documents = splitter.split_documents(docs)

    embedding = HuggingFaceEmbeddings(
        model_name="BAAI/bge-small-en-v1.5"
    )

    vectorstore = FAISS.from_documents(documents, embedding)

    retriever = vectorstore.as_retriever(
        search_type="similarity",
        search_kwargs={"k":3}
    )

    placement_data = create_retriever_tool(
        retriever=retriever,
        name="placement_data",
        description="Search the placement database for company eligibility, job role, job description, CGPA, skills and hiring information."
    )

    llm = ChatGroq(
        model="llama-3.3-70b-versatile",
        temperature=0,
        api_key="YOUR KEY"
    )

    agent = create_react_agent(
        model=llm,
        tools=[placement_data]
    )

    system_prompt = """
You are an expert Placement Assistant.

Your job is to answer ONLY using the information retrieved from the
placement database.

Instructions:
1. Always use the placement_data tool whenever the question is about companies, placements, CGPA, skills, eligibility, hiring, job roles, or job descriptions.
2. Never make up company information.
3. If the information is not available in the database, reply:
   "I couldn't find that information in the placement database."
4. Give concise and well-formatted answers.
5. If company information exists, present it as:

Company:
Job Role:
Minimum CGPA:
Required Skills:
Job Description:

6. If multiple companies match, list each separately.
7. Do not answer unrelated questions.
"""

    response = agent.invoke(
        {
            "messages": [
                SystemMessage(content=system_prompt),
                HumanMessage(content=query)
            ]
        }
    )

    return response["messages"][-1].content
````

````python
import os
import requests
from bs4 import BeautifulSoup
from ddgs import DDGS
def event_agent(query):

    # ==========================
    # Groq API
    # ==========================
    os.environ["GROQ_API_KEY"] = "YOUR KEY"

    llm = ChatGroq(
        model_name="llama-3.3-70b-versatile",
        temperature=0
    )

    # ==========================
    # Search Events
    # ==========================
    results = []

    with DDGS() as ddgs:
        for r in ddgs.text(query, max_results=5):
            results.append({
                "title": r["title"],
                "url": r["href"],
                "body": r["body"]
            })

    if len(results) == 0:
        return "No events found."

    final_answer = []

    # ==========================
    # Process Every Event
    # ==========================
    for r in results:

        try:

            headers = {
                "User-Agent": "Mozilla/5.0"
            }

            response = requests.get(
                r["url"],
                headers=headers,
                timeout=10
            )

            soup = BeautifulSoup(
                response.text,
                "html.parser"
            )

            page = soup.get_text(
                separator=" ",
                strip=True
            )[:8000]

        except Exception:
            continue

        if len(page) < 300:
            continue

        prompt = f"""
You are an event assistant.

Extract ONLY the following:

Event Name
Date
Venue
Fee
Eligibility
Registration Deadline
Organizer
Registration Link (if present)
Short Description

If any field is missing write:

Not Mentioned

Website Content:

{page}
"""

        summary = llm.invoke(prompt).content

        final_answer.append(
            f"""
{'='*80}
TITLE:
{r['title']}

{summary}

SOURCE:
{r['url']}
{'='*80}
"""
        )

    if len(final_answer) == 0:
        return "No useful events found."

    return "\n".join(final_answer)
````

````python
def college_knowledge_agent(query, pdf_path="/content/AR22-CSE-SYLLABUS 2023-24.pdf"):
    import os

    from langchain_community.document_loaders import PyMuPDFLoader
    from langchain_text_splitters import RecursiveCharacterTextSplitter
    from langchain_community.embeddings import HuggingFaceEmbeddings
    from langchain_community.vectorstores import FAISS
    from langchain_groq import ChatGroq

    # ================================
    # Groq API Key
    # ================================
    os.environ["GROQ_API_KEY"] = "YOUR KEY"

    llm = ChatGroq(
        model="llama-3.3-70b-versatile",
        temperature=0
    )

    # ================================
    # Load PDF
    # ================================
    loader = PyMuPDFLoader(pdf_path)
    documents = loader.load()

    # ================================
    # Split Documents
    # ================================
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=200
    )

    docs = splitter.split_documents(documents)

    # ================================
    # Create Embeddings
    # ================================
    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2"
    )

    # ================================
    # Create FAISS Vector Store
    # ================================
    vectorstore = FAISS.from_documents(
        docs,
        embeddings
    )

    # ================================
    # FAISS Retriever
    # ================================
    retriever = vectorstore.as_retriever(
        search_type="similarity",
        search_kwargs={"k": 5}
    )

    # ================================
    # Retrieve Documents
    # ================================
    retrieved_docs = retriever.invoke(query)

    context = "\n\n".join(doc.page_content for doc in retrieved_docs)

    # ================================
    # Prompt
    # ================================
    prompt = f"""
You are an AR22 Regulation Assistant.

Answer ONLY using the retrieved context.

Rules:
1. Never hallucinate.
2. If the answer is unavailable, reply exactly:
   "I couldn't find that information in the AR22 regulations."
3. If multiple retrieved chunks contain relevant information, combine them.

Context:
{context}

Question:
{query}

Answer:
"""

    answer = llm.invoke(prompt).content

    return answer
````

If the requested information is not present in the document, the agent responds:

> **"I couldn't find that information in the document."**

This ensures reliable, accurate, and trustworthy responses based solely on the official transportation handbook.

````python
def transport_agent(query, pdf_path="/content/transport_agent_x_final.pdf"):

    import os
    import fitz
    import numpy as np
    from scipy.sparse import csr_matrix
    from sklearn.feature_extraction.text import TfidfVectorizer
    from sklearn.metrics.pairwise import cosine_similarity
    from langchain_groq import ChatGroq

    # ==========================
    # Hardcode your Groq API Key
    # ==========================
    os.environ["GROQ_API_KEY"] = "YOUR KEY"

    # ==========================
    # Load PDF
    # ==========================
    doc = fitz.open(pdf_path)

    pages = [
        {
            "page": i + 1,
            "text": page.get_text()
        }
        for i, page in enumerate(doc)
    ]

    documents = [p["text"] for p in pages]

    # ==========================
    # TF-IDF
    # ==========================
    vectorizer = TfidfVectorizer(stop_words="english")

    tfidf_matrix = vectorizer.fit_transform(documents)

    # ==========================
    # Positional Encoding
    # ==========================
    d_model = tfidf_matrix.shape[1]

    def positional_encoding(position):

        pe = np.zeros(d_model)

        for i in range(0, d_model, 2):

            div_term = np.power(10000.0, i / d_model)

            pe[i] = np.sin(position / div_term)

            if i + 1 < d_model:
                pe[i + 1] = np.cos(position / div_term)

        return pe

    pe_matrix = np.vstack(
        [positional_encoding(i) for i in range(len(pages))]
    )

    weighted_matrix = tfidf_matrix + csr_matrix(pe_matrix)

    # ==========================
    # Retrieval
    # ==========================
    query_vector = vectorizer.transform([query])

    similarities = cosine_similarity(
        query_vector,
        weighted_matrix
    )[0]

    top_k = 20

    top_indices = similarities.argsort()[::-1][:top_k]

    context = "\n\n".join(
        pages[idx]["text"]
        for idx in top_indices
    )

    # ==========================
    # LLM
    # ==========================
    llm = ChatGroq(
        model_name="llama-3.3-70b-versatile",
        temperature=0
    )

    prompt = f"""
You are an expert document question-answering assistant.

Answer ONLY from the provided context.

If the answer is unavailable, reply exactly:

I couldn't find that information in the document.

Context:
{context}

Question:
{query}

Answer:
"""

    answer = llm.invoke(prompt).content

    return answer
````

# 🧠 Hybrid Multi-Agent Intent Router

## 📌 Overview

The **Hybrid Multi-Agent Intent Router** is the decision-making component of the Campus AI Assistant. Its primary responsibility is to intelligently analyze every user query and route it to the most appropriate specialized agent.

Instead of relying on a single routing strategy, the system adopts a **three-stage hybrid cascade architecture**, progressively increasing computational complexity only when necessary. This approach minimizes latency, reduces LLM usage, and improves routing accuracy.

The routing pipeline consists of:

1. **TF-IDF Semantic Matching**
2. **Sentence Embedding Similarity**
3. **Groq LLM Intent Classification (Fallback)**

This hybrid design ensures that simple queries are routed quickly while complex or ambiguous queries are resolved using an LLM.

---

# 🎯 Objectives

- Route user queries to the correct specialized agent.
- Minimize unnecessary LLM calls.
- Improve routing accuracy.
- Reduce inference latency.
- Support ambiguous and natural language queries.
- Build a scalable multi-agent architecture.

# 🚀 Routing Pipeline

The routing process follows three sequential stages.

### Stage 1 — TF-IDF Router

The incoming query is converted into a TF-IDF vector and compared with predefined agent descriptions.

Similarity is computed using:

- TF-IDF Vectorizer
- Cosine Similarity

If the confidence exceeds the predefined threshold (**0.60**), the corresponding agent is immediately selected.

**Advantages**

- Extremely fast
- Lightweight
- No LLM cost
- Excellent for keyword-based queries

Example

```
Query:
TCS eligibility

↓

Placement Agent
```

---

### Stage 2 — Sentence Embedding Router

If TF-IDF confidence is insufficient, the router generates dense semantic embeddings using the **BAAI/bge-small-en-v1.5** embedding model.

Cosine similarity is computed between:

- User query embedding
- Agent description embeddings

If semantic confidence exceeds **0.30**, the matching agent is selected.

**Advantages**

- Understands semantic meaning
- Handles synonyms
- Better than keyword matching

Example

```
Query:
Companies requiring Java skills

↓

Placement Agent
```

Although "eligibility" is not explicitly mentioned, semantic similarity correctly identifies the placement domain.

---

### Stage 3 — Groq LLM Router

If both previous stages produce low confidence, the query is forwarded to a Groq-hosted LLM.

The model performs intent classification by selecting exactly one category:

- Placement
- Questions
- Project
- Event
- College
- Transport

The returned category determines the final specialized agent.

**Advantages**

- Handles ambiguous queries
- Supports conversational language
- Highest routing accuracy

Example

```
Query:
Can I participate?

↓

Event Agent
```

The LLM understands that the user is referring to an event registration.

---

# 📊 Confidence Strategy

Instead of selecting the highest cosine similarity directly, the router applies **temperature-scaled Softmax confidence**.

Benefits include:

- Better probability estimation
- More reliable confidence scores
- Improved threshold decisions
- Reduced routing ambiguity

Thresholds used:

| Routing Tier | Threshold |
|--------------|-----------|
| TF-IDF | 0.60 |
| Sentence Embedding | 0.30 |
| LLM | Fallback |

````python
# ============================================================
# Multi-agent intent router — HYBRID CASCADE (implementation)
# TF-IDF tier -> Sentence-embedding tier -> Groq LLM tier
# Confidence rule: SOFTMAX confidence (temperature-scaled)
# Exposes: route_query(query) -> agent_name
#          handle_query(query) -> final answer (dispatches to
#          your air/water/noise/energy/soil functions)
# ============================================================

import os
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer
from groq import Groq

# ---------------------------------------------------------------------------
# Groq client
# ---------------------------------------------------------------------------
# NOTE: pull this from your .env / environment instead of hardcoding it.
os.environ["GROQ_API_KEY"] = "YOUR KEY"
groq_client = Groq(api_key=os.environ.get("GROQ_API_KEY"))
GROQ_MODEL = "llama-3.1-8b-instant"

# ---------------------------------------------------------------------------
# Agents
# ---------------------------------------------------------------------------
# ---------------------------------------------------------------------------
# Agents
# ---------------------------------------------------------------------------

AGENT1 = "placement_agent"
AGENT2 = "important_questions_agent"
AGENT3 = "project_idea_agent"
AGENT4 = "event_agent"
AGENT5 = "college_knowledge_agent"
AGENT6 = "transport_agent"

AGENTS = [
    AGENT1,
    AGENT2,
    AGENT3,
    AGENT4,
    AGENT5,
    AGENT6,
]

descriptions = [

"""
Expert in campus placements and career guidance.

Handles queries related to company eligibility, minimum CGPA,
required technical skills, aptitude preparation, interview rounds,
job descriptions, internship opportunities, placement statistics,
recruitment process, resume preparation, hiring criteria,
coding rounds, campus drives, salary packages,
and placement-related information.
""",

"""
Expert in academic question generation.

Handles requests to generate important questions,
exam questions, viva questions, MCQs,
short-answer questions, long-answer questions,
assignment questions, previous-year style questions,
interview questions, unit-wise questions,
chapter-wise questions, topic-wise questions,
and exam preparation material from uploaded notes,
PDFs, or study resources.
""",

"""
Expert in project ideas and innovation.

Handles requests for software projects,
AI projects, Machine Learning projects,
Deep Learning projects, Data Science projects,
Cybersecurity projects, IoT projects,
Cloud Computing projects, Web Development projects,
Mobile App projects, Blockchain projects,
Final Year Projects, Mini Projects,
Research Ideas, Project Architecture,
Technology Stack recommendations,
Datasets, APIs, and implementation guidance.
""",

"""
Expert in college events and activities.

Handles queries related to workshops,
seminars, hackathons, coding competitions,
technical events, cultural events,
sports events, registrations,
event schedules, venues,
student club activities,
fest information,
certification programs,
guest lectures,
and event announcements.
""",

"""
Expert in college information and academics.

Handles questions related to departments,
faculty members, HODs,
principal, administration,
academic regulations,
college rules,
fee structure,
admissions,
hostel facilities,
library,
laboratories,
clubs,
academic calendar,
semester information,
timetables,
contact information,
campus facilities,
and general college knowledge.
""",

"""
Expert in college transportation services.

Handles queries related to bus routes,
bus timings,
transport schedules,
pickup locations,
drop locations,
bus numbers,
transport fees,
hostel transport,
college shuttle services,
vehicle tracking,
driver information,
transport office,
parking facilities,
and transportation-related services.
"""
]

AGENT_LABEL = {
    AGENT1: "placement",
    AGENT2: "questions",
    AGENT3: "project",
    AGENT4: "event",
    AGENT5: "college",
    AGENT6: "transport",
}

LABEL_TO_AGENT = {v: k for k, v in AGENT_LABEL.items()}

# ---------------------------------------------------------------------------
# Models / vectors (built once, at import time)
# ---------------------------------------------------------------------------
sen = SentenceTransformer("BAAI/bge-small-en-v1.5")

tfidf = TfidfVectorizer()
doc_vectors = tfidf.fit_transform(descriptions)
doc_embeddings = sen.encode(descriptions)

# ---- Confidence rule: SOFTMAX (temperature-scaled) ----
TEMPERATURE = 0.1
TFIDF_THRESHOLD = 0.60   # softmax confidence threshold for TF-IDF tier
ST_THRESHOLD = 0.30      # softmax confidence threshold for sentence tier

def softmax_confidence(scores):
    scores = np.array(scores)
    p = np.exp(scores / TEMPERATURE)
    p /= p.sum()
    idx = p.argmax()
    confidence = p.max()
    return idx, confidence

# ---------------------------------------------------------------------------
# Tier 1: TF-IDF
# ---------------------------------------------------------------------------
def _tfidf_tier(query: str):
    vec = tfidf.transform([query])
    scores = cosine_similarity(vec, doc_vectors)[0]
    idx, confidence = softmax_confidence(scores)
    return idx, confidence

# ---------------------------------------------------------------------------
# Tier 2: Sentence embeddings
# ---------------------------------------------------------------------------
def _sentence_tier(query: str):
    emb = sen.encode([query])
    scores = cosine_similarity(emb, doc_embeddings)[0]
    idx, confidence = softmax_confidence(scores)
    return idx, confidence

# ---------------------------------------------------------------------------
# Tier 3: Groq LLM fallback
# ---------------------------------------------------------------------------
def _llm_tier(query: str) -> str:
    system_prompt = """
You are an intelligent router.

Your job is to classify the user's query into exactly ONE category.

Categories:

placement
- placements
- companies
- internship
- eligibility
- CGPA
- interview
- coding round
- hiring
- job role
- resume
- skills
- salary package

questions
- important questions
- viva
- MCQs
- assignments
- PYQs
- notes
- exam preparation
- question generation
- chapter questions
- unit questions

project
- project ideas
- AI project
- ML project
- IoT project
- Cybersecurity project
- Web project
- Research idea
- Final year project
- Mini project

event
- workshop
- seminar
- hackathon
- event
- competition
- fest
- registration
- cultural event
- sports event

college
- department
- faculty
- HOD
- principal
- college rules
- admissions
- hostel
- fee structure
- timetable
- library
- laboratory
- clubs

transport
- bus
- transport
- bus timing
- pickup
- drop
- shuttle
- route
- parking
- transport fee

Return ONLY one word from:

placement
questions
project
event
college
transport

Do not explain.
Do not use punctuation.
"""
    try:
        response = groq_client.chat.completions.create(
            model=GROQ_MODEL,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": query},
            ],
            temperature=0,
            max_tokens=5,
        )
        label = response.choices[0].message.content.strip().lower()
        label = "".join(ch for ch in label if ch.isalpha())
        return LABEL_TO_AGENT.get(label, AGENT1)  # safe default if parsing fails
    except Exception as e:
        print(f"[llm_tier] Groq call failed, defaulting to {AGENT1}: {e}")
        return AGENT1

# ---------------------------------------------------------------------------
# Tier usage counters (optional, for monitoring)
# ---------------------------------------------------------------------------
tfidf_count = 0
st_count = 0
llm_count = 0

# ---------------------------------------------------------------------------
# Main entry point: hybrid cascade
# ---------------------------------------------------------------------------
def route_query(query: str) -> str:
    """
    Route a single query to one of the 5 agents using a hybrid cascade:
      1. TF-IDF cosine similarity + softmax confidence
      2. Sentence-embedding cosine similarity + softmax confidence
      3. Groq LLM classification (fallback)
    Returns the agent name (e.g. 'agent1_water').
    """
    global tfidf_count, st_count, llm_count

    idx, confidence = _tfidf_tier(query)
    if confidence > TFIDF_THRESHOLD:
        tfidf_count += 1
        return AGENTS[idx]

    idx, confidence = _sentence_tier(query)
    if confidence > ST_THRESHOLD:
        st_count += 1
        return AGENTS[idx]

    llm_count += 1
    return _llm_tier(query)

# ---------------------------------------------------------------------------
# Dispatch table: agent name -> your existing topic function
# (air, water, noise, energy, soil must already be defined above this cell)
# ---------------------------------------------------------------------------
AGENT_FUNC_MAP = {
    AGENT1: placement_agent,
    AGENT2: important_questions_agent,
    AGENT3: project_idea_agent,
    AGENT4: event_agent,
    AGENT5: college_knowledge_agent,
    AGENT6: transport_agent,
}

def handle_query(query: str) -> str:
    """
    Full pipeline:
      1. route_query() picks the right agent via TF-IDF -> sentence -> LLM cascade
      2. AGENT_FUNC_MAP dispatches to the matching topic function
      3. That function runs its own Groq chain and returns the answer
    """
    agent_name = route_query(query)
    func = AGENT_FUNC_MAP.get(agent_name)

    if func is None:
        return f"No handler found for agent '{agent_name}'."

    print(f"[router] '{query}' -> {agent_name} -> {func.__name__}()")
    return func(query)
````

````python
import faiss
from langchain_community.vectorstores import FAISS
from langchain_community.docstore import InMemoryDocstore
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_core.documents import Document

# Embedding model
embedding = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# FAISS Index
index = faiss.IndexFlatL2(384)

# Cache Vector Store
QA_CACHE = FAISS(
    embedding_function=embedding,
    index=index,
    docstore=InMemoryDocstore({}),
    index_to_docstore_id={}
)
````

````python
def cache_answer(state):

    result = QA_CACHE.similarity_search_with_relevance_scores(
        state.query,
        k=1
    )

    if not result:
        state.boole = False
        return state

    doc, score = result[0]

    if score >= 0.95:
        state.enhanced_answer = doc.metadata["answer"]
        state.boole = True      # Cache Hit
    else:
        state.boole = False     # Cache Miss

    return state
````

````python
import gradio as gr

# -------------------------------
# Backend Function
# -------------------------------
def ask(query):
    result = app.invoke(
    {
        "query": query,
        "enhanced_query": "",
        "selected_agent": "",
        "answer": "",
        "boole":False,
        "enhanced_answer": ""

    }
)
    return result["enhanced_answer"]


def think():
    """Shown immediately on submit, before the real answer resolves."""
    return "_Consulting the record…_"


def fill_topic(topic):
    return topic


# -------------------------------
# Design system
# -------------------------------
# Concept: "The Reading Room" — a scholar's desk, not a chatbot.
# Cool paper background, ink-navy text, a single brass/gold seal accent
# used sparingly (the header monogram + the answer's margin rule).
#
# Color    ink #1C2430 · indigo #223A5E · brass #A97C33 · slate #626C7A
#          hairline #DDE1E6 · paper #F3F4F1 · card #FFFFFF
# Type     display/body serif "Source Serif 4" · UI sans "IBM Plex Sans"
#          utility mono "IBM Plex Mono" (small caps labels)

custom_css = """
@import url('https://fonts.googleapis.com/css2?family=Source+Serif+4:ital,wght@0,400;0,500;0,600;1,400&family=IBM+Plex+Sans:wght@400;500;600&family=IBM+Plex+Mono:wght@500&display=swap');

:root{
    --ink:#1C2430;
    --indigo:#223A5E;
    --brass:#A97C33;
    --brass-light:#C79A52;
    --slate:#626C7A;
    --hairline:#DDE1E6;
    --paper:#F3F4F1;
    --card:#FFFFFF;
}

*{ font-family:'IBM Plex Sans', sans-serif; }

body, .gradio-container{
    background:var(--paper) !important;
}

.gradio-container{
    max-width:100% !important;
    min-height:100vh;
    padding:0 !important;
}

footer{ display:none !important; }

/* ---------- Header ---------- */

#header{
    background:var(--card);
    border-bottom:1px solid var(--hairline);
    padding:26px 32px;
    position:sticky;
    top:0;
    z-index:100;
}

.header-content{
    max-width:820px;
    margin:auto;
    display:flex;
    align-items:center;
    gap:16px;
}

.seal{
    flex:none;
    width:44px;
    height:44px;
    border-radius:50%;
    border:1.5px solid var(--brass);
    display:flex;
    align-items:center;
    justify-content:center;
    font-family:'Source Serif 4', serif;
    font-weight:600;
    font-size:16px;
    color:var(--brass);
    letter-spacing:0.5px;
}

.header-text{ flex:1; }

.title{
    font-family:'Source Serif 4', serif;
    font-size:24px;
    font-weight:600;
    color:var(--ink);
    letter-spacing:0.2px;
}

.subtitle{
    margin-top:4px;
    color:var(--slate);
    font-size:13.5px;
    font-style:italic;
    font-family:'Source Serif 4', serif;
}

/* ---------- Answer ---------- */

#answer-column{
    max-width:820px;
    margin:auto;
    padding:40px 24px 28px 24px;
}

#response-label{
    font-family:'IBM Plex Mono', monospace;
    font-size:11px;
    font-weight:500;
    letter-spacing:2px;
    text-transform:uppercase;
    color:var(--brass);
    margin-bottom:10px;
    padding-left:20px;
}

#answer{
    background:var(--card) !important;
    border:1px solid var(--hairline) !important;
    border-left:3px solid var(--brass) !important;
    border-radius:4px !important;
    padding:26px 28px !important;
    box-shadow:0 12px 28px rgba(28,36,48,.05);
    min-height:220px;
}

#answer, #answer *{
    color:#1C2430 !important;
    -webkit-text-fill-color:#1C2430 !important;
}

#answer .prose{
    font-family:'Source Serif 4', serif !important;
    font-size:16.5px !important;
    line-height:1.85 !important;
}

#answer .prose p:first-child{ margin-top:0; }
#answer .prose em{ color:#626C7A !important; -webkit-text-fill-color:#626C7A !important; }

/* ---------- Quick topics ---------- */

#topics{
    max-width:820px;
    margin:8px auto 0 auto;
    padding:0 24px;
    display:flex;
    gap:8px;
    flex-wrap:wrap;
}

#topics button{
    background:var(--card) !important;
    border:1px solid var(--hairline) !important;
    color:var(--indigo) !important;
    font-size:12.5px !important;
    font-weight:500 !important;
    border-radius:100px !important;
    padding:6px 16px !important;
    box-shadow:none !important;
    min-width:0 !important;
}

#topics button:hover{
    border-color:var(--brass) !important;
    color:var(--brass) !important;
}

/* ---------- Input ---------- */

#input-area{
    position:sticky;
    bottom:0;
    background:linear-gradient(180deg, rgba(243,244,241,0) 0%, var(--paper) 40%);
    padding:22px 18px 22px 18px;
}

#input-row{
    max-width:820px;
    margin:auto;
    background:var(--card);
    border:1px solid var(--hairline);
    border-radius:16px;
    padding:8px 8px 8px 20px;
    box-shadow:0 10px 30px rgba(28,36,48,.08);
    display:flex;
    align-items:center;
    transition:border-color .15s ease, box-shadow .15s ease;
}

#input-row:focus-within{
    border-color:var(--brass);
    box-shadow:0 0 0 3px rgba(169,124,51,.12);
}

#ask-label{
    font-family:'IBM Plex Mono', monospace;
    font-size:10px;
    font-weight:500;
    letter-spacing:1.5px;
    color:var(--slate);
    text-transform:uppercase;
}

#question textarea{
    border:none !important;
    background:transparent !important;
    box-shadow:none !important;
    font-size:15px !important;
    padding:10px 8px !important;
    font-family:'IBM Plex Sans', sans-serif !important;
    color:#1C2430 !important;
    -webkit-text-fill-color:#1C2430 !important;
}

#question textarea::placeholder{
    color:#8890A0 !important;
    -webkit-text-fill-color:#8890A0 !important;
}

#send{
    background:var(--indigo) !important;
    color:white !important;
    border:none !important;
    border-radius:12px !important;
    height:44px !important;
    min-width:44px !important;
    width:44px !important;
    padding:0 !important;
    font-weight:600 !important;
    transition:background .15s ease;
}

#send:hover{ background:var(--brass) !important; }

#footer{
    text-align:center;
    color:var(--slate);
    font-family:'IBM Plex Mono', monospace;
    font-size:10.5px;
    letter-spacing:1px;
    text-transform:uppercase;
    margin-top:14px;
    opacity:0.7;
}
"""

# -------------------------------
# UI
# -------------------------------

with gr.Blocks(
    theme=gr.themes.Soft(
        primary_hue="blue",
        secondary_hue="slate",
        neutral_hue="gray",
        font=gr.themes.GoogleFont("IBM Plex Sans"),
    ),
    css=custom_css,
    title="Academic Assistant"
) as demo:

    gr.HTML("""
    <div id="header">
        <div class="header-content">
            <div class="seal">AA</div>
            <div class="header-text">
                <div class="title">Academic Assistant</div>
                <div class="subtitle">
                    Ask about regulations, curriculum, credits, examinations and academic policy.
                </div>
            </div>
        </div>
    </div>
    """)

    with gr.Column(elem_id="answer-column"):
        gr.HTML('<div id="response-label">Response</div>')
        answer = gr.Markdown(
            elem_id="answer",
            value="_Ask a question below to begin — for example, how credit hours are calculated, "
                  "or the policy on re-examinations._"
        )

    topic_credits = None  # placeholder to keep names local to scope below

    with gr.Row(elem_id="topics"):
        t1 = gr.Button("Credit requirements", size="sm")
        t2 = gr.Button("Examination policy", size="sm")
        t3 = gr.Button("Curriculum structure", size="sm")
        t4 = gr.Button("Attendance rules", size="sm")

    with gr.Column(elem_id="input-area"):

        with gr.Row(elem_id="input-row"):

            question = gr.Textbox(
                elem_id="question",
                placeholder="Ask your question…",
                lines=1,
                max_lines=6,
                show_label=False,
                container=False,
                scale=9,
                autofocus=True
            )

            send = gr.Button("→", elem_id="send", scale=0)

        gr.HTML("""
        <div id="footer">Powered by LangGraph · Groq · FAISS</div>
        """)

    send.click(fn=think, inputs=None, outputs=answer).then(fn=ask, inputs=question, outputs=answer)
    question.submit(fn=think, inputs=None, outputs=answer).then(fn=ask, inputs=question, outputs=answer)

    t1.click(lambda: "What are the minimum credit requirements to graduate?", None, question)
    t2.click(lambda: "What is the policy on re-examinations?", None, question)
    t3.click(lambda: "How is the curriculum structured across semesters?", None, question)
    t4.click(lambda: "What are the attendance requirements?", None, question)

demo.launch()
````
